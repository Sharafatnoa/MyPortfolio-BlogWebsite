It should let you set `f(x, y) = 0.5 * (x**2 + 8 * y**2)`, pick a learning rate, and plot the descent path over the contour lines shown on the post.

```python
import numpy as np

def f(x, y):
    return 0.5 * (x**2 + 8 * y**2)

def grad(x, y):
    return np.array([x, 8 * y])

def descend(start, lr, steps=40):
    path = [np.array(start, dtype=float)]
    for _ in range(steps):
        path.append(path[-1] - lr * grad(*path[-1]))
    return np.array(path)

descend((-4.4, 1.55), lr=0.1)
```
